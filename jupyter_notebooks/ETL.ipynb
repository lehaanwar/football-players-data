As there are a lot of columns for the skills, I will create new columns below to help sparse the data, which will help clean the data. So it will be grouped by defensive, attacking, speed, technical ability and passing

ML Feature engineering - Manual new feature

I have now added new columns to group certain skills together, I will now use ML to drop some columns that do not need to be there. 

```python
# import libararies ML feature engineering
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from feature_engine.selection import DropFeatures, SmartCorrelatedSelection
from feature_engine.encoding import OrdinalEncoder
```

```python
print(df.columns.to_list())
```

```python
# create a new feature called 'defense_attributes' by taking the mean of the following columns: marking', 'standing_tackle', 'sliding_tackle', 'aggression', 'strength', 'interceptions', 'composure'
df['defense_attributes'] = df[['marking', 'standing_tackle', 'sliding_tackle', 'aggression', 'strength', 'interceptions', 'composure']].mean(axis=1)

# create a new feature called attack_attributes by taking the mean of the following columns: 'finishing', 'heading_accuracy', 'volleys', 'long_shots', 'shot_power', 'positioning', 'penalties'
df['attack_attributes'] = df[['finishing', 'heading_accuracy', 'volleys', 'long_shots', 'shot_power', 'positioning', 'penalties']].mean(axis=1)

# create a new feature called passing_attributes by taking the mean of the following columns: short_passing, long_passing, vision, curve, freekick_accuracy, crossing, ball_control, vision, reactions, dribbling
df['passing_attributes'] = df[['short_passing', 'long_passing', 'vision', 'curve', 'freekick_accuracy', 'crossing', 'ball_control', 'vision', 'reactions', 'dribbling']].mean(axis=1)

# create a new feature called pace_attributes by taking the mean of the following columns: acceleration, sprint_speed, agility, balance, stamina, jumping
df['pace_attributes'] = df[['acceleration', 'sprint_speed', 'agility', 'balance', 'stamina', 'jumping']].mean(axis=1)

# create a pipeline for feature engineering
pipeline = Pipeline([
    ('drop_features', DropFeatures(features_to_drop=['name', 'birth_date', 'national_rating', 'national_team_position', 'national_jersey_number', 'crossing', 'finishing', 'heading_accuracy', 'short_passing', 'volleys', 'dribbling', 'curve', 'freekick_accuracy', 'long_passing', 'ball_control', 'acceleration', 'sprint_speed', 'agility', 'reactions', 'balance', 'shot_power', 'jumping', 'stamina', 'strength', 'long_shots', 'aggression', 'interceptions', 'positioning', 'vision', 'penalties', 'composure', 'marking', 'standing_tackle', 'sliding_tackle']))
])

# after setting up the pipeline, we can fit and transform the data
df = pipeline.fit_transform(df)
df.head()
```

The columns that were not required were dropped, making the data easier to analyse. I was going to do a smart correlated feature to check which groups correlate with each other but I made the decision to not add this due to time and maybe it would show high correlation with my nw groups and the overall rating.
