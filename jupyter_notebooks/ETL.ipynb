I tried a boxplot of all numerical data but there were too many columns so I wanted to check the skills of the players as these are shown from 0-100. 

```python
# show outliers for the players skils; crossing, finishing, heading_accuracy, short_passing, volleys,dribbling, curve,freekick_accuracy, long_passing, ball_control, acceleration, sprint_speed, agility, reactions, balance	shot_power, jumping, stamina, strength, long_shots, aggression, interceptions, positioning, vision, penalties, composure, marking, standing_tackle, sliding_tackle
# List of skills to check for outliers 0-100
skills = ['crossing', 'finishing', 'heading_accuracy', 'short_passing', 'volleys', 'dribbling', 'curve', 'freekick_accuracy', 'long_passing', 'ball_control', 'acceleration', 'sprint_speed', 'agility', 'reactions', 'balance', 'shot_power', 'jumping', 'stamina', 'strength', 'long_shots', 'aggression', 'interceptions', 'positioning', 'vision', 'penalties', 'composure', 'marking', 'standing_tackle', 'sliding_tackle']
for skill in skills:
    plt.figure(figsize=(10, 5))
    sns.boxplot(x=df[skill])
    plt.title(f'Boxplot of {skill}')
    plt.show()
```

The above shows some columns have 0-20 and some have 80-100, I think they should be fine as this could be the goalkeepers. I will go into a deeper analysis of this by splitting the goalkeepers into a column of their own

```python
# check the outliers for position for goalkeepers in skills; crossing, finishing, heading_accuracy, short_passing, volleys,dribbling, curve,freekick_accuracy, long_passing, ball_control, acceleration, sprint_speed, agility, reactions, balance	shot_power, jumping, stamina, strength, long_shots, aggression, interceptions, positioning, vision, penalties, composure, marking, standing_tackle, sliding_tackle

# List of skill columns (0-100 range)
skill_cols = [
    'crossing', 'finishing', 'heading_accuracy', 'short_passing', 'volleys',
    'dribbling', 'curve', 'freekick_accuracy', 'long_passing', 'ball_control',
    'acceleration', 'sprint_speed', 'agility', 'reactions', 'balance',
    'shot_power', 'jumping', 'stamina', 'strength', 'long_shots',
    'aggression', 'interceptions', 'positioning', 'vision', 'penalties',
    'composure', 'marking', 'standing_tackle', 'sliding_tackle',
    'overall_rating', 'potential'
]

# Filter to only columns that exist in your dataframe
skill_cols = [col for col in skill_cols if col in df.columns]

# Plot
fig, axes = plt.subplots(nrows=8, ncols=4, figsize=(20, 30))
axes = axes.flatten()

for i, col in enumerate(skill_cols):
    sns.boxplot(y=df[col], ax=axes[i], color='steelblue')
    axes[i].set_title(col, fontsize=10)
    axes[i].set_xlabel('')

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Boxplots of Skill Columns', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```
